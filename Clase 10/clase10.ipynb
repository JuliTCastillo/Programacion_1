<p><b>a) ¿Qué significa que la información persista en el tiempo?</b></p>
<p>Esto quiere decir que se consevar en la información guardada más alla del programa que la creo/ejecuto.</p>
<p><b>b) ¿Qué diferencias encuentran entre una variable en memoria y un archivo?</b></p>
<p>Las variables en memoria se van creado a medida que un programa lo solicita y mediante el flujo del programa vamos cargando la información, una vez que el programa termina, su información se pierde. En cambio, cuando usamos un archivo, esto se guarda en el disco es un medio de almacenamiento, si el programa termina, el archivo se sigue conservando con la información que se guardo en el momento (no se pierde).</p>
<p><b>c) ¿Qué ventajas y limitaciones presenta almacenar los datos como texto?</b></p>
<p>La ventaja de guardar los datos como texto, es que es mas legible para cualquier editor y se puede leer los datos directamente, sin necesidad de programas externos, podemos detectar más rápido algunos fallos, ya que podemos acceder al archivo en un vista simple. Sin embargo, tenemos algunas limitaciones porque pueden ocupar más espacio que tal vez un archivo binario, debemos tener en cuenta la conversión de los datos (de un str a int) y su busqueda es de datos es más lenta, ya que debemos leer constantemente el archivo completo.
</p>

<h4>Actividad 2</h4>
<p><b>a) "rt": lectura; el archivo debe existir.</b></p>

```python
with open("notas.txt", "rt", encoding="utf-8") as archivo:
    print(archivo.read())
```

<p><b>b) "wt": escritura; crea el archivo o reemplaza todo su contenido si ya existe.</b></p>

```python
with open("notas.txt", "wt", encoding="utf-8") as archivo:
    archivo.write("Primer registro\n")
    archivo.write("Segundo registro\n")
    archivo.write("Tercer registro\n")
```

<p><b>c) "at": agregado; crea el archivo o escribe al final sin borrar lo anterior</b></p>

```python
with open("notas.txt", "at", encoding="utf-8") as archivo:
    archivo.write("Cuarto registro\n")
```

<p>La función <code>open()</code> se usa para abrir el archivo, indicando el modo (lectura, escritura, agregado) y el tipo en el que sistema traducen los caracteres de texto a bytes (utf-8).</p>
<p>La función <code>write()</code> es para indicar que vamos a escribir sobre el archivo, esto siempre y cuando tengamos el modo <b>"wr" o "at"</b>. Si se intenta hacer un <code>write()</code> estando en el modo <b>"rt"</b> el programa nos va a tirar error.</p>
<p><code>with</code> garantiza que el archivo se cierre SIEMPRE (incluso si hay un errro), esto pasas adentro de un bloque. En el caso de que no usemos el <code>with</code> al finalizar el uso del archivo deberiamos poner <code>archivo.close()</code> para que se cierre.</p>

<h4>Actividad 3</h4>

```python
def validadcion(dato):
    valor = ""
    if dato == "legajo":
        try:
            valor = int(input("Ingrese el numero de legajo: "))
            while valor < 0:
                print(f"El {dato} debe ser un numero positivo.")
                valor = int(input("Ingrese el numero de legajo: "))
        except ValueError as err:
            print(f"Error: {err}")

    elif dato == "nombre" or dato == "apellido":
        valor = input(f"Ingrese el {dato} del alumno: ")
        while len(valor) == 0:
            print(f"El {dato} esta vacio.")
            valor = input(f"Ingrese el {dato} del alumno: ")

    elif dato == "comision":
        while True:
            try:
                valor = int(input("Ingrese la comisión (1 a 10): "))
                if 1 <= valor <= 10:
                    break
                else:
                    print("La comisión debe ser un número entre 1 y 10.")
            except ValueError:
                print("La comisión debe ser un número.")

    return valor
    
def crear_alumno(legajo, nombre, apellido, comision):
    try:
        arch = open("alumnos.txt", "at", encoding="utf-8")
    except IOError as err:
        print(f"Error: {err}")
    else: 
        arch.write(f"{legajo};{nombre};{apellido};{comision}\n")
    finally:
        arch.close()

def main():
    legajo = validadcion("legajo")  
    while legajo != -1:     
        nombre = validadcion("nombre")
        apellido = validadcion("apellido")
        comision = validadcion("comision")

        crear_alumno(legajo, nombre, apellido, comision)

        legajo = int(input("Ingrese el legajo del alumno: "))

main()
```

<p>Si en la función de <code>crear_alumno</code> usamos el modo <b>"wt"</b> la información se va pisando. Por otro lado, al usar el modo <b>"at"</b> la información se va agregando con la que ya hay en el archivo txt.</p>

<h4>Actividad 5: productos.csv</h4><p>Se usa punto y coma como delimitador y no se utiliza el modulo csv. Las funciones omiten el encabezado, validan los registros, calculan importes y avisan si hay menos de cinco unidades.</p>

```python
encabezado_productos = "codigo;descripcion;cantidad;precio\n"

def cargar_productos():
    try:
        with open("productos.csv", "wt", encoding="utf-8") as archivo:
            archivo.write(encabezado_productos)
            while True:
                codigo = input("Codigo (FIN para terminar): ").strip()
                if codigo.upper() == "FIN":
                    break
                if not codigo or ";" in codigo:
                    print("Codigo vacio o con delimitador invalido.")
                    continue
                
                descripcion = input("Descripcion: ").strip()
                if not descripcion or ";" in descripcion:
                    print("Descripcion vacia o con delimitador invalido.")
                    continue

                try:
                    cantidad = int(input("Cantidad (>= 0): "))
                    precio = float(input("Precio (> 0): "))
                except ValueError:
                    print("Cantidad y precio deben ser numericos.")
                    continue

                if cantidad < 0 or precio <= 0:
                    print("La cantidad debe ser >= 0 y el precio > 0.")
                    continue

                archivo.write(f"{codigo};{descripcion};{cantidad};{precio}\n")
    except OSError as error:
        print("No se pudo escribir productos.csv:", error)

def leer_productos():
    total = 0.0
    try:
        with open("productos.csv", "rt", encoding="utf-8") as archivo:
            encabezado = archivo.readline().rstrip("\n")
            if encabezado != encabezado_productos.rstrip("\n"):
                print("Encabezado ausente o incorrecto.")
                return

            nro = 2
            for linea in archivo:
                campos = linea.rstrip("\n").split(";")
                if len(campos) != 4:
                    print(f"Linea {nro} invalida: se esperaban 4 campos.")
                    nro += 1
                    continue

                codigo, descripcion, cant_txt, precio_txt = campos
                try:
                    cantidad, precio = int(cant_txt), float(precio_txt)
                except ValueError:
                    print(f"Linea {nro} invalida: cantidad o precio incorrecto.")
                    nro += 1
                    continue

                if not codigo or not descripcion or cantidad < 0 or precio <= 0:
                    print(f"Linea {nro} invalida: datos vacios o fuera de rango.")
                    nro += 1
                    continue

                importe = cantidad * precio
                total += importe
                print(f"{codigo} - {descripcion}: importe ${importe:.2f}")
                if cantidad < 5:
                    print("  Cantidad menor a 5")

                nro += 1
    except OSError as error:
        print("No se pudo leer productos.csv:", error)
        return

    print(f"Importe total: ${total:.2f}")
```

<h4>Actividad 6</h4><p>Los ejercicios manejan <code>OSError</code> al abrir, crear o escribir archivos; <code>ValueError</code> solo alrededor de conversiones numericas. Las filas mal formadas se informan y omiten para continuar. <code>with</code> cierra el archivo aunque ocurra una excepcion.</p>

<h4>Actividad 7</h4>

```python
encabezado_entregas = "grupo;proyecto;responsable;avance\n"

def crear_archivo_entregas():
    try:
        with open("entregas.csv", "wt", encoding="utf-8") as archivo:
            archivo.write(encabezado_entregas)
    except OSError as error:
        print("No se pudo crear entregas.csv:", error)

def leer_entregas():
    entregas = []
    try:
        with open("entregas.csv", "rt", encoding="utf-8") as archivo:
            encabezado = archivo.readline().rstrip("\n")
            if encabezado != encabezado_entregas.rstrip("\n"):
                print("Encabezado de entregas.csv ausente o incorrecto.")
                return entregas

            nro = 2
            for linea in archivo:
                campos = linea.rstrip("\n").split(";")
                if len(campos) != 4:
                    print(f"Linea {nro} invalida: se esperaban 4 campos.")
                    nro += 1
                    continue

                grupo_txt, proyecto, responsable, avance_txt = campos
                try:
                    grupo = int(grupo_txt)
                    avance = float(avance_txt)
                except ValueError:
                    print(f"Linea {nro} invalida: grupo o avance no numerico.")
                    nro += 1
                    continue

                if grupo <= 0 or not proyecto.strip() or not responsable.strip() or not (0 <= avance <= 100):
                    print(f"Linea {nro} invalida: campos vacios o fuera de rango.")
                    nro += 1
                    continue
                
                entregas.append((grupo, proyecto, responsable, avance))
                nro += 1
    except OSError as error:
        print("No se pudo acceder a entregas.csv:", error)

    return entregas

def cargar_entregas():
    
    entregas_existentes = leer_entregas()
    grupos = set()
    for e in entregas_existentes:
        grupos.add(e)

    try:
        with open("entregas.csv", "at", encoding="utf-8") as archivo:
            while True:
                try:
                    grupo = int(input("Numero de grupo (0 para terminar): "))
                except ValueError:
                    print("El grupo debe ser entero.")
                    continue

                if grupo == 0:
                    break

                if grupo < 0 or grupo in grupos:
                    print("El grupo debe ser positivo y no estar repetido.")
                    continue

                proyecto = input("Nombre del proyecto: ").strip()
                responsable = input("Nombre del responsable: ").strip()

                if not proyecto or ";" in proyecto or not responsable or ";" in responsable:
                    print("Proyecto o responsable invalido (no vacio y sin ';').")
                    continue

                try:
                    avance = float(input("Avance (0 a 100): "))
                except ValueError:
                    print("El avance debe ser numerico.")
                    continue

                if not (0 <= avance <= 100):
                    print("El avance debe estar entre 0 y 100.")
                    continue

                archivo.write(f"{grupo};{proyecto};{responsable};{avance}\n")
                grupos.add(grupo)
                print(f"Grupo {grupo} cargado exitosamente.")
    except OSError as error:
        print("No se pudo escribir entregas.csv:", error)

def mostrar_entregas():
    entregas = leer_entregas()
    for grupo, proyecto, responsable, avance in entregas:
        print(f"Grupo {grupo}: {proyecto} | {responsable} | {avance:.1f}%")

def buscar_entrega(grupo_buscado):
    entregas = leer_entregas()
    for grupo, proyecto, responsable, avance in entregas:
        if grupo == grupo_buscado:
            print(f"Grupo {grupo}: {proyecto} | {responsable} | {avance:.1f}%")
            return (grupo, proyecto, responsable, avance)

    print(f"No se encontro el grupo {grupo_buscado}.")
    return None

def informar_avances_bajos():
    entregas = leer_entregas()
    for grupo, proyecto, responsable, avance in entregas:
        if avance < 50:
            print(f"Grupo {grupo}: {proyecto}, {avance:.1f}% ({responsable})")

def promedio_avance():
    entregas = leer_entregas()
    if not entregas:
        print("No hay entregas validas para calcular el promedio.")
        return None

    suma = 0
    for grupo, proyecto, responsable, avance in entregas:
        suma += avance

    promedio = suma / len(entregas)
    print(f"Promedio de avance: {promedio:.2f}%")
    return promedio
```
