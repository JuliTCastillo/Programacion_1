<h4>Actividad 1</h4>
<p><b>a) ¿Qué significa que la información persista en el tiempo?</b></p>
<p>Esto quiere decir que se consevar en la información guardada más alla del programa que la creo/ejecuto.</p>
<p><b>b) ¿Qué diferencias encuentran entre una variable en memoria y un archivo?</b></p>
<p>Las variables en memoria se van creado a medida que un programa lo solicita y mediante el flujo del programa vamos cargando la información, una vez que el programa termina, su información se pierde. En cambio, cuando usamos un archivo, esto se guarda en el disco es un medio de almacenamiento, si el programa termina, el archivo se sigue conservando con la información que se guardo en el momento (no se pierde).</p>
<p><b>c) ¿Qué ventajas y limitaciones presenta almacenar los datos como texto?</b></p>
<p>La ventaja de guardar los datos como texto, es que es mas legible para cualquier editor y se puede leer los datos directamente, sin necesidad de programas externos, podemos detectar más rápido algunos fallos, ya que podemos acceder al archivo en un vista simple. Sin embargo, tenemos algunas limitaciones porque pueden ocupar más espacio que tal vez un archivo binario, debemos tener en cuenta la conversión de los datos (de un str a int) y su busqueda es de datos es más lenta, ya que debemos leer constantemente el archivo completo.
</p>

<h4>Actividad 2</h4>
<p><b>a) "rt": lectura; el archivo debe existir.</b></p>

```python
with open("notas.txt", "rt", encoding="utf-8") as archivo:
    print(archivo.read())
```

<p><b>b) "wt": escritura; crea el archivo o reemplaza todo su contenido si ya existe.</b></p>

```python
with open("notas.txt", "wt", encoding="utf-8") as archivo:
    archivo.write("Primer registro\n")
    archivo.write("Segundo registro\n")
    archivo.write("Tercer registro\n")
```

<p><b>c) "at": agregado; crea el archivo o escribe al final sin borrar lo anterior</b></p>

```python
with open("notas.txt", "at", encoding="utf-8") as archivo:
    archivo.write("Cuarto registro\n")
```

<p>La función <code>open()</code> se usa para abrir el archivo, indicando el modo (lectura, escritura, agregado) y el tipo en el que sistema traducen los caracteres de texto a bytes (utf-8).</p>
<p>La función <code>write()</code> es para indicar que vamos a escribir sobre el archivo, esto siempre y cuando tengamos el modo <b>"wr" o "at"</b>. Si se intenta hacer un <code>write()</code> estando en el modo <b>"rt"</b> el programa nos va a tirar error.</p>
<p><code>with</code> garantiza que el archivo se cierre SIEMPRE (incluso si hay un errro), esto pasas adentro de un bloque. En el caso de que no usemos el <code>with</code> al finalizar el uso del archivo deberiamos poner <code>archivo.close()</code> para que se cierre.</p>

<h4>Actividad 3</h4>

```python
def validadcion(dato):
    valor = ""
    if dato == "legajo":
        try:
            valor = int(input("Ingrese el numero de legajo: "))
            while valor < 0:
                print(f"El {dato} debe ser un numero positivo.")
                valor = int(input("Ingrese el numero de legajo: "))
        except ValueError as err:
            print(f"Error: {err}")

    elif dato == "nombre" or dato == "apellido":
        valor = input(f"Ingrese el {dato} del alumno: ")
        while len(valor) == 0:
            print(f"El {dato} esta vacio.")
            valor = input(f"Ingrese el {dato} del alumno: ")

    elif dato == "comision":
        while True:
            try:
                valor = int(input("Ingrese la comisión (1 a 10): "))
                if 1 <= valor <= 10:
                    break
                else:
                    print("La comisión debe ser un número entre 1 y 10.")
            except ValueError:
                print("La comisión debe ser un número.")

    return valor
    
def crear_alumno(legajo, nombre, apellido, comision):
    try:
        arch = open("alumnos.txt", "at", encoding="utf-8")
    except IOError as err:
        print(f"Error: {err}")
    else: 
        arch.write(f"{legajo};{nombre};{apellido};{comision}\n")
    finally:
        arch.close()

def main():
    legajo = validadcion("legajo")  
    while legajo != -1:     
        nombre = validadcion("nombre")
        apellido = validadcion("apellido")
        comision = validadcion("comision")

        crear_alumno(legajo, nombre, apellido, comision)

        legajo = int(input("Ingrese el legajo del alumno: "))

main()
```

<p>Si en la función de <code>crear_alumno</code> usamos el modo <b>"wt"</b> la información se va pisando. Por otro lado, al usar el modo <b>"at"</b> la información se va agregando con la que ya hay en el archivo txt.</p>
