<p>Antes de comenzar, respondan en grupo:</p>
<p>a) ¿Qué diferencia existe entre problema, algoritmo y programa?</br>
<p><strong>Diferencia entre problema, algoritmo y programa:</strong></p>
<p>
Un <strong>problema</strong> es una situación o necesidad que se desea resolver .<br>
Un <strong>algoritmo</strong> es la secuencia finita y ordenada de pasos lógicos que describe cómo resolver ese problema, independientemente del lenguaje de programación .<br>
Un <strong>programa</strong> es la implementación concreta de ese algoritmo escrita en un lenguaje de programación específico, para que la computadora pueda ejecutarlo.
</p>
b) ¿Qué ventaja ofrece dividir un programa en funciones?</br>
<p>
Dividir un programa en funciones permite:<br>
- Reutilizar código sin tener que reescribirlo.<br>
- Facilitar la lectura y el mantenimiento del programa.<br>
- Dividir un problema grande en subproblemas más pequeños y manejables (modularidad).<br>
- Facilitar las pruebas y la detección de errores, ya que cada función se puede probar por separado.<br>
- Permitir el trabajo en equipo, dado que distintas personas pueden encargarse de distintas funciones.
</p>
c) ¿Qué diferencia existe entre mostrar un resultado y retornarlo?</br>
<p><strong>Diferencia entre mostrar un resultado y retornarlo:</strong></p>
<p>
<strong>Mostrar</strong> un resultado (por ejemplo con <code>print</code>) significa imprimirlo por pantalla para que el usuario lo vea, pero ese valor no queda disponible para ser usado por el programa.<br>
<strong>Retornar</strong> un resultado (con <code>return</code>) significa que la función devuelve ese valor a quien la llamó, quedando disponible para almacenarlo en una variable o utilizarlo en otras partes del programa.
</p>
d) ¿Cómo se accede, modifica y recorre una lista?</br>
<p><strong>Cómo se accede, modifica y recorre una lista:</strong></p>
<p>
<strong>Acceder:</strong> se utiliza el índice de la posición entre corchetes, por ejemplo <code>lista[0]</code> accede al primer elemento .<br>
<strong>Modificar:</strong> se asigna un nuevo valor a una posición específica, por ejemplo <code>lista[2] = 10</code>.<br>
<strong>Recorrer:</strong> se utiliza un bucle, generalmente <code>for elemento in lista:</code> para recorrer cada elemento, o <code>for i in range(len(lista)):</code> cuando se necesita también el índice.
</p>
</p>

<p>
Ya que <code>"Fortnite"</code> existe en la lista, <code>buscar_juego()</code> retorna <strong>2</strong>, su índice dentro de la lista. Con <code>"Zelda"</code> que no existe, retorna <strong>-1</strong>, ya que el <code>while</code> recorre toda la lista sin encontrar coincidencia y la variable <code>posicion</code> nunca cambia de su valor inicial.
</p>

<p><strong>c) ¿Por qué una función puede modificar una lista recibida como parámetro?</strong></p>
<p>
Porque en Python las listas son objetos <strong>mutables</strong> y se pasan por <strong>referencia</strong>: el parámetro <code>juegos</code> dentro de <code>agregar_juego()</code> no es una copia, sino que apunta al mismo objeto en memoria que la lista original definida en <code>principal.py</code>. Por eso, cuando se ejecuta <code>juegos.append(titulo)</code> dentro de la función, ese cambio se ve reflejado también afuera, en la lista original.
</p>

<h4><strong>Verificacion y Analisis</strong></h4>
<h5>Explicación: funciones que retornan valores y funciones que modifican listas</h5>

<p><strong>Funciones que solamente retornan un valor sin modificar la lista:</strong></p>
<p>
- <code>buscar_juego(juegos, titulo)</code>: recorre la lista con un <code>while</code> y <strong>retorna</strong> un número entero: la posición del título si lo encuentra, o -1 si no existe. En ningún momento cambia el contenido de <code>juegos</code>.<br>
- <code>cantidad_juegos(juegos)</code>: <strong>retorna</strong> un número entero (<code>len(juegos)</code>) con la cantidad de elementos. Tampoco modifica la lista, solo consulta su longitud.
</p>

<p><strong>Función que modifica la lista recibida y además retorna un valor:</strong></p>
<p>
- <code>agregar_juego(juegos, titulo)</code>: es la única de las cuatro que <strong>modifica</strong> la lista original, usando <code>juegos.append(titulo)</code> cuando el título no existe. Esto es posible porque las listas se pasan por referencia. Además de modificarla, también <strong>retorna</strong> un valor booleano (<code>True</code> o <code>False</code>) para indicar si pudo agregar el juego o no.
</p>

<p><strong>Función que no retorna nada ni modifica la lista:</strong></p>
<p>
- <code>mostrar_juegos(juegos)</code>: solo recorre la lista con un <code>for</code> e imprime cada elemento por pantalla. No modifica <code>juegos</code> ni tiene ningún <code>return</code>, por lo que su valor de retorno es <code>None</code>.
</p>

<h5><strong>Reflexión</strong></h5>

<p>
De las herramientas vistas en la clase (slicing, funciones lambda y comprensión de listas), la que más mejoró la <strong>claridad</strong> y la <strong>reutilización</strong> del código fue el <strong>slicing</strong>. Nos permitio extraer o modificar partes específicas de una lista en una sola línea, sin necesidad de escribir bucles con <code>for</code> o <code>while</code> ni variables auxiliares para ir guardando resultados. Esto hace que el código sea más corto, más legible de un vistazo, y más fácil de reutilizar en distintas partes del programa, ya que una expresión como <code>lista[::2]</code> o <code>lista[::-1]</code> resume en pocos caracteres una operación que de otra forma llevaría varias líneas.
</p>
