<h4>La primera hora de la clase resolverán el siguiente problema:</h4> 
<p>La empresa de turismo GUADALAJARA VIAJES comercializa 10 diferentes tours a través de 15 agencias en todo el país. Diariamente cada agencia envía la cantidad de ventas efectuadas de cada tour, indicando agencia, tours y la cantidad de pasajes. Algunas agencias envían 0 en la cantidad cuando no existen pasajes para un tour determinado, pero no en todas. </br></br>
La información que se envía tiene los siguientes datos que se ingresan en forma desordenada: </br>
• Código de agencia (de 1 a 15) </br>
• Código de tour (1 a 10) </br>
• Cantidad de pasajes (0 a 20) </br></br>
Informar: </br></br>
a. El total de pasajes vendidos por tour en cada agencia. </br>
b. El número de agencia que obtuvo la máxima cantidad de pasajes vendidos.</p>

```python
import random

def cargarDatos(matriz):
    for i in range(15):
        matriz.append([i+1, random.randint(1,10), random.randint(0,20)])
        
def cantidadDePasaje(matriz):
    for i in range(len(matriz)):
        print("En la cantidad de pasajes vendidos en la agencia", matriz[i][0], "fue de", matriz[i][2], "pasajes.")

def maximoPasaje(matriz):
    max = -9999
    nAgencia = -1
    for i in range(len(matriz)):
        if matriz[i][2] > max:
            max = matriz[i][2]
            nAgencia = matriz[i][0]

    print("La agencia número", nAgencia, "vendio la mayor cantidad de pasajes de un total de", max) 
def main():
    matrizViaje = []
    cargarDatos(matrizViaje)
    cantidadDePasaje(matrizViaje)
    print()
    maximoPasaje(matrizViaje)

main()
```
