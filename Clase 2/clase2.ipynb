<p>Antes de comenzar, respondan en grupo:</p>
<p>a) ¿Qué diferencia existe entre problema, algoritmo y programa?</br>
b) ¿Qué ventaja ofrece dividir un programa en funciones?</br>
c) ¿Qué diferencia existe entre mostrar un resultado y retornarlo?</br>
d) ¿Cómo se accede, modifica y recorre una lista?</br>
</p>

<p>a)</br>Problema: Es la situación o necesidad que se requiere para resolver algo.</p>
<p>Algoritmo: Es el conjunto de pasos ordenados que indican cómo se resuelve el problema.</p>
<p>Programa: Es la implementación de el algoritmo mediante un lenguaje de programación para que la computadora pueda ejecutarlo.</p>

<p>b) La ventaja de dividir un programa en funciones permite organizar mejor el código, reutilizar la funciones, facilitar la comprensión y hacer más sencilla la detección de errores.</p>


<p>c) <b>Mostrar</b> un resultado en pantalla solo nos puede servir para informar el dato o poder visualizar un resultado, pero no tenes acceso a esa información, ya que se pierde. En cambio, <b>retornar</b> un resultado, nos permite guardarlo en memoria para poder utilizarlo en algún momento del programa.</p>

<p>d)
Acceder: se utiliza su índice. Ejemplo: lista[0].</br>
Modificar: se asigna un nuevo valor a una posición. Ejemplo: lista[0] = 10.</br>
Recorrer: se utiliza normalmente un ciclo for. Ejemplo: for elemento in lista: print(elemento).</br>
</p>
