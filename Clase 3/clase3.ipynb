<p><b>a) ¿Qué ventajas ofrece separar un programa en módulos?</b></p>
<p>Separar un programa en módulos permite organizar mejor el código, hacerlo más fácil de entender, mantener y modificar. También permite reutilizar funciones y facilita encontrar y corregir errores.</p>

<p><b>b) ¿Qué diferencia existe entre una lista y una cadena de caracteres?</b></p>
<p>Una lista puede contener varios elementos de diferentes tipos y se puede modificar. Una cadena de caracteres es una secuencia de caracteres utilizada para representar texto y es inmutable.</p>

<p><b>c) ¿Qué significa que un dato sea mutable o inmutable?</b></p>
<p>Un dato mutable puede modificarse después de haber sido creado, como una lista. Un dato inmutable no puede modificarse una vez creado, como una cadena de caracteres o una tupla.</p>

<p><b>d) ¿Qué responsabilidades conviene dejar en el programa principal y cuáles en las funciones?</b></p>
<p>En el programa principal conviene dejar el control general del programa y la organización de las acciones. En las funciones conviene colocar tareas específicas, como cálculos, validaciones o procesamiento de datos, para que puedan reutilizarse y mantener el código ordenado.</p>

<p><b>2. Git y GitHub: conceptos fundamentales</b></br></br>
a) <b>Repositorio local</b>: Lugar donde git almacena el proyecto y su historia de cambios en la computadora.</br>
b) <b>Repositorio remoto</b>: Repositorio alojado en un servidor, como GitHub, que permite compartir y sincronizar el proyecto.</br>
c) <b>Área de preparación o staging</b>: Espacio donde se seleccionan los cambios que se quieren incluir en el próximo commit.</br>
d) <b>Commit</b>: Registro de un conjunto de cambios realizados en el proyecto, creando un punto en el historial.</br>
e) <b>Push y pull</b>: <b>push</b> envía los cambios locales al repositorio remoto, mientras que el <b>pull</b> obtiene los cambios del repositorio remoto.</br>
</p>

<p><b>a) ¿Qué información muestra <code>git status</code>?</b></br></br>
<p><code>git status</code> muestra el estado actual del repositorio. Indica qué archivos fueron modificados, cuáles están preparados para el commit y cuáles todavía no fueron agregados al área de preparación.</p>

<p><b>¿Por qué <code>isalpha()</code> retorna False cuando hay espacios?</b></p>

<p><code>isalpha()</code> retorna <b>True</b> solamente cuando todos los caracteres de la cadena son letras. Los espacios no son letras, por eso una cadena como <code>"Programacion en Python"</code> retorna <code>False</code>, aunque todas las palabras que contiene estén formadas únicamente por letras.</p>

<p>De manera similar, <code>isalnum()</code> verifica que todos los caracteres sean letras o números. Como los espacios tampoco son letras ni números, también retorna <code>False</code> en este caso.</p>

<p><b>7.Conversión de tipos y formateo</b></p>

<p>El programa solicita el nombre del producto, el precio unitario y la cantidad. Luego convierte el precio a <code>float</code>, la cantidad a <code>int</code> y calcula el importe total.</p>


```python
producto = input("Producto: ")
precio = float(input("Precio unitario: "))
cantidad = int(input("Cantidad: "))

total = precio * cantidad

# a) Utilizando f-string
print(f"Producto: {producto} | Total: ${total:.2f}")

# b) Utilizando concatenación
print("Producto: " + producto + " | Total: $" + str(total))
```


<p><b>a) F-string</b></p>

<p>La f-string permite insertar directamente variables dentro del texto y utilizar <code>:.2f</code> para mostrar el total con exactamente dos decimales.</p>

<p>Por ejemplo, si se ingresa:</p>

<ul>
    <li>Producto: <code>Mouse</code></li>
    <li>Precio unitario: <code>15000</code></li>
    <li>Cantidad: <code>2</code></li>
</ul>

<p>El resultado es:</p>

<pre><code>Producto: Mouse | Total: $30000.00</code></pre>

<p><b>b) Concatenación</b></p>

<p>Al utilizar concatenación, los valores numéricos no se pueden unir directamente con cadenas, por lo que es necesario convertirlos explícitamente utilizando <code>str()</code>.</p>

<pre><code>print("Producto: " + producto + " | Total: $" + str(total))</code></pre>

<p>El resultado sería:</p>

<pre><code>Producto: Mouse | Total: $30000.0</code></pre>

<p><b>Comparación</b></p>

<ul>
    <li><b>F-string:</b> es más clara y fácil de leer.</li>
    <li><b>Concatenación:</b> requiere convertir explícitamente los valores numéricos con <code>str()</code>.</li>
    <li><b>Formato:</b> la f-string permite mostrar fácilmente dos decimales utilizando <code>:.2f</code>.</li>
</ul>

<p><b>Conclusión:</b> las f-strings son más recomendables porque permiten escribir el resultado de forma más simple, legible y con un mejor control del formato.</p>
