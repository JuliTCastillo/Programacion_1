<p><b>2. Git y GitHub: conceptos fundamentales</b></br></br>
a) <b>Repositorio local</b>: Lugar donde git almacena el proyecto y su historia de cambios en la computadora.</br>
b) <b>Repositorio remoto</b>: Repositorio alojado en un servidor, como GitHub, que permite compartir y sincronizar el proyecto.</br>
c) <b>Área de preparación o staging</b>: Espacio donde se seleccionan los cambios que se quieren incluir en el próximo commit.</br>
d) <b>Commit</b>: Registro de un conjunto de cambios realizados en el proyecto, creando un punto en el historial.</br>
e) <b>Push y pull</b>: <b>push</b> envía los cambios locales al repositorio remoto, mientras que el <b>pull</b> obtiene los cambios del repositorio remoto.</br>
</p>

<p><b>4. Primer commit y trabajo colaborativo seguro</b></p>


<p><b>a) ¿Qué información muestra git status?</b></br></br>
Muestra el estado actual del repositorio, indicando qué archivos fueron modificados, cuáles son nuevos, cuáles están preparados para el commit y cuáles todavía no fueron agregados.
</p>

<p><b>b) ¿Qué diferencia existe entre git add y git commit?</b></br></br>
<code>git add</code> Selecciona y prepara los cambios que queremos incluir y el <code>git commit</code> registra esos cambios preparados en el historial del repositorio.
</p>

<p><b>c) ¿Por qué cada integrante debe ejecutar git pull antes de comenzar?</b></br></br>
Para obtener los últimos cambios realizados por los demás integrantes y trabajar sobre una versión actualizada del proyecto.
</p>
