## 2. Git y GitHub: conceptos fundamentales

- **Repositorio local:** carpeta de la computadora donde Git guarda el historial de versiones del proyecto.
- **Repositorio remoto:** copia del repositorio alojada en un servidor, como GitHub, que permite compartir el proyecto con otros.
- **Staging:** área intermedia donde se colocan los cambios antes de confirmarlos con un commit.
- **Commit:** registro de una versión del proyecto en un momento determinado, acompañado por un mensaje descriptivo.
- **Push:** sube los commits locales al repositorio remoto.
- **Pull:** trae los cambios del repositorio remoto a la copia local.

## 6. Funciones y métodos de cadenas

Algunos métodos útiles para trabajar con cadenas son `upper()`, `lower()`, `title()`, `capitalize()`, `replace()`, `count()`, `find()` y `len()`.

````python
texto = "Programacion en Python"

print("upper():", texto.upper())
print("lower():", texto.lower())
print("title():", texto.title())
print("capitalize():", texto.capitalize())
print("replace():", texto.replace("Python", "IA"))
print("count():", texto.count("a"))
print("find():", texto.find("Python"))
print("len():", len(texto))
````

### Métodos de validación

`isalpha()` verifica si todos los caracteres son letras, `isdigit()` verifica si todos son dígitos e `isalnum()` verifica si todos son letras o números.

En una cadena que contiene espacios, `isalpha()` devuelve `False` porque el espacio no es una letra.

````python
cadenas = ["Programacion", "2026", "Python3", "Programacion en Python"]

for cadena in cadenas:
    print("Cadena:", cadena)
    print("isalpha():", cadena.isalpha())
    print("isdigit():", cadena.isdigit())
    print("isalnum():", cadena.isalnum())
    print()
````

## 7. Conversión de tipos y formateo

El siguiente programa solicita el nombre de un producto, su precio unitario y la cantidad. El precio se convierte a `float` y la cantidad a `int`. Luego se calcula el importe total y se muestra mediante una f-string y mediante concatenación.

````python
nombre_producto = input("Ingrese el nombre del producto: ")
precio_unitario = float(input("Ingrese el precio unitario: "))
cantidad = int(input("Ingrese la cantidad: "))

importe_total = precio_unitario * cantidad

# a) Utilizando una f-string
print(f"Producto: {nombre_producto} - Importe total: ${importe_total:.2f}")

# b) Utilizando concatenación
print("Producto: " + nombre_producto + " - Importe total: $" + str(round(importe_total, 2)))
````

### Comparación

La f-string es más legible y permite formatear directamente los valores, por ejemplo con `.2f` para mostrar dos decimales. Con concatenación es necesario convertir explícitamente los valores numéricos con `str()`, por lo que el código resulta más largo y puede ser más propenso a errores.

## 8. Actividad integradora: perfil del equipo

El programa solicita el nombre del equipo, la comisión, el nombre de cada integrante y su rol inicial. Los nombres se normalizan con `title()`, el nombre del equipo se convierte a mayúsculas, se informa su cantidad de caracteres, se genera una sigla y se verifica si contiene al menos un dígito.

````python
def contiene_digitos(texto):
    for caracter in texto:
        if caracter.isdigit():
            return True
    return False


def generar_sigla(nombre_equipo):
    palabras = nombre_equipo.split()
    sigla = ""
    for palabra in palabras:
        sigla += palabra[0].upper()
    return sigla


# Entrada de datos
nombre_equipo = input("Ingrese el nombre del equipo: ")
comision = input("Ingrese la comisión: ")

cantidad_integrantes = int(input("Ingrese la cantidad de integrantes: "))
integrantes = []

for i in range(cantidad_integrantes):
    nombre = input(f"Ingrese el nombre del integrante {i + 1}: ").title()
    rol = input(f"Ingrese el rol inicial de {nombre}: ")
    integrantes.append((nombre, rol))

# Procesamiento
nombre_equipo = nombre_equipo.upper()
sigla = generar_sigla(nombre_equipo)
cantidad_caracteres = len(nombre_equipo)
tiene_digitos = contiene_digitos(nombre_equipo)

# Salida
print(f"\nEquipo: {nombre_equipo}")
print(f"Comisión: {comision}")
print(f"Cantidad de caracteres: {cantidad_caracteres}")
print(f"Sigla: {sigla}")
print(f"¿Contiene algún dígito?: {tiene_digitos}")

print("\nIntegrantes:")
for nombre, rol in integrantes:
    print(f"- {nombre}: {rol}")
````
