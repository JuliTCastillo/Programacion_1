## 1. Repaso de la clase anterior

**a) ¿Qué ventajas ofrece separar un programa en módulos?**
Permite organizar el código en partes más chicas y reutilizables, facilita el mantenimiento, hace más fácil encontrar errores y permite que varias personas trabajen en distintas partes sin pisarse.

**b) ¿Qué diferencia existe entre una lista y una cadena de caracteres?**
Ambas son secuencias, pero la lista es mutable (se puede modificar después de creada) y puede contener elementos de distintos tipos. La cadena es inmutable y solo contiene caracteres.

**c) ¿Qué significa que un dato sea mutable o inmutable?**
Mutable significa que se puede modificar su contenido después de creado sin cambiar su identidad (como las listas). Inmutable significa que una vez creado no se puede alterar; cualquier "cambio" en realidad crea un objeto nuevo (como las cadenas y las tuplas).

**d) ¿Qué responsabilidades conviene dejar en el programa principal y cuáles en las funciones?**
En el programa principal conviene dejar la entrada y salida de datos (pedir datos al usuario, mostrar resultados). En las funciones conviene dejar el procesamiento y la lógica de cálculo, para que sean reutilizables y fáciles de testear por separado.

## 2. Git y GitHub: conceptos fundamentales

Relación de cada concepto con su función:

- **a) Repositorio local** → copia del proyecto que vive en la computadora, donde se trabaja y se hacen los commits.
- **b) Repositorio remoto** → copia alojada en GitHub, compartida entre todos los integrantes.
- **c) Área de preparación (staging)** → zona intermedia donde se colocan los cambios con `git add` antes de confirmarlos con un commit.
- **d) Commit** → registro/foto de los cambios preparados, guardado en el historial del repositorio local.
- **e) Push y pull** → `push` sube los commits locales al repositorio remoto; `pull` trae los cambios del remoto a la copia local.

## 5. Cadenas de caracteres

Una cadena es una secuencia inmutable de caracteres.

```python
texto = "Programacion en Python"

# a) Primer y último carácter
print(texto[0])
print(texto[-1])

# b) Obtener "Programacion" mediante slicing
print(texto[0:12])

# c) Cadena invertida
print(texto[::-1])

# d) Verificar si "Python" pertenece a la cadena
print("Python" in texto)

# e) Intentar modificar texto[0]
texto[0] = "p"
```

**e) Explicación del error:**
Al ejecutar `texto[0] = "p"` Python arroja un `TypeError: 'str' object does not support item assignment`, porque las cadenas son inmutables: no se puede reemplazar un carácter por índice, hay que crear una cadena nueva (por ejemplo con slicing o `replace()`).

## 6. Funciones y métodos de cadenas

Los métodos que transforman una cadena retornan una nueva cadena; no modifican el objeto original.

```python
texto = "Programacion en Python"

print(texto.upper())
print(texto.lower())
print(texto.title())
print(texto.capitalize())
print(texto.replace("Python", "IA"))
print(texto.count("a"))
print(texto.find("Python"))
print(len(texto))
```

Porque `isalpha()` exige que todos los caracteres de la cadena sean letras. El espacio no es una letra, entonces aunque todas las palabras estén compuestas por letras, la presencia del espacio hace que el chequeo completo falle.

## 7. Conversión de tipos y formateo

Programa que solicita nombre del producto, precio unitario y cantidad, y calcula el importe total.

```python
producto = input("Producto: ")
precio = float(input("Precio unitario: "))
cantidad = int(input("Cantidad: "))
total = precio * cantidad

# a) f-string con 2 decimales
print(f"Total a pagar por {cantidad} {producto}: ${total:.2f}")

# b) Concatenación con str()
print("Total a pagar por " + str(cantidad) + " " + producto + ": $" + str(round(total, 2)))
```

**Comparación:**
La f-string es más legible y no requiere conversiones manuales (Python las hace solas dentro de `{}`), mientras que con concatenación hay que convertir explícitamente cada valor numérico con `str()`, lo que hace el código más largo y propenso a errores si te olvidás de convertir algo.

## 8. Actividad integradora: perfil del equipo

Programa perfil_equipo.py que solicita nombre del equipo, comisión, nombre de cada integrante y rol inicial.

```python
def normalizar_nombre(nombre):
    return nombre.title()

def generar_sigla(nombre_equipo):
    palabras = nombre_equipo.split()
    return "".join(palabra[0].upper() for palabra in palabras)

def contiene_digitos(texto):
    val = False
    for caracter in texto:
        if caracter.isdigit():
            val = True
    return val

# Programa principal
nombre_equipo = input("Nombre del equipo: ")
comision = input("Comisión: ")
cantidad_integrantes = int(input("Cantidad de integrantes: "))

nombre_equipo_mayus = nombre_equipo.upper()
sigla = generar_sigla(nombre_equipo)
tiene_digitos = contiene_digitos(nombre_equipo)

print(f"\nEquipo: {nombre_equipo_mayus}")
print(f"Comisión: {comision}")
print(f"Cantidad de caracteres del nombre: {len(nombre_equipo)}")
print(f"Sigla: {sigla}")
print(f"¿El nombre contiene al menos un dígito?: {tiene_digitos}")

for i in range(cantidad_integrantes):
    nombre = input(f"Nombre del integrante {i+1}: ")
    rol = input(f"Rol inicial de {nombre}: ")
    print(f"- {normalizar_nombre(nombre)} | Rol: {rol}")
```
