## 5. Operadores, funciones y métodos



```python
ventas = (120, 85, 230, 150, 90, 150)

# a) Cantidad, mayor, menor y suma total
print("Cantidad:", len(ventas))
print("Mayor:", max(ventas))
print("Menor:", min(ventas))
print("Suma total:", sum(ventas))

# b) Verificar pertenencia de 150 y ausencia de 500
print(150 in ventas)
print(500 not in ventas)

# c) Cantidad de veces que aparece 150
print(ventas.count(150))

# d) Primera posición de 230
print(ventas.index(230))

# e) Concatenar ventas con (300, 250)
nueva_tupla = ventas + (300, 250)
print(nueva_tupla)
print(type(nueva_tupla))

# f) Replicar (0, 1) tres veces
print((0, 1) * 3)
```

```python
# El método index() genera ValueError si el elemento no está presente
valor = 999
if valor in ventas:
    print(ventas.index(valor))
else:
    print(f"{valor} no está presente en la tupla")
```

**Punto e):** al sumar dos tuplas con `+`, Python no modifica ninguna de las originales (son inmutables), sino que crea y devuelve una tupla nueva. Se puede confirmar imprimiendo `ventas` nuevamente y viendo que no cambió.

**Punto f):** el operador `*` con una tupla repite sus elementos la cantidad de veces indicada, por eso `(0, 1) * 3` da `(0, 1, 0, 1, 0, 1)`.

```python
codigos = ("P101", "P205", "P330") 
ventas_semanales = [ 
    [12, 15, 10, 18], 
    [8, 11, 9, 14], 
    [20, 17, 22, 19] ] 
```

<p><b>a) Informar el total vendido por producto.</b></p>

```python
def cant_vendidosXcodigo(posicion, ventas):
    total = sum(ventas[posicion])
    return total

for codigo in codigos:
    print(f"El producto {codigo} vendio un total de {cant_vendidosXcodigo(codigos.index(codigo), ventas_semanales)}")
```

<p><b>b) Informar el total de una semana indicada.</b></p>

```python
def cant_vendidosXsemana(semana,ventas):
    total = 0
    for producto in ventas:
        total += producto[semana]
    return total

semana = 1
print(f"Se vendio un total de {cant_vendidosXsemana(semana,ventas_semanales)} durante la semana numero {semana}")
```

<p><b>c) Determinar el código del producto con mayor venta acumulada.</b></p>

```python
def cant_vendidosXcodigo(posicion,ventas):
    total = sum(ventas[posicion])
    return total

def codigoXventas(codigos,ventas):
    resultados = []
    for codigo in codigos:
        ventas_totales = cant_vendidosXcodigo(codigos.index(codigo),ventas)
        resultados.append((codigo,ventas_totales))
    return resultados

def mayor_ventas(tabla_ventas):
    codigo = tabla_ventas[0][0]
    mayor = tabla_ventas[0][1]
    for i in range(1,len(tabla_ventas)):
        if tabla_ventas[i][1] > mayor:
            codigo = tabla_ventas[i][0]
            mayor = tabla_ventas[i][1]
    print(f"El producto que mas vendio fue el {codigo} con {mayor} ventas")

mayor_ventas(codigoXventas(codigos,ventas_semanales))
```

<p><b>d) Validar la semana ingresada y capturar los errores correspondientes.</b></p>

```python
def pedir_semana(ventas):
    entrada = input("Ingrese el número de semana (0 a 3): ")
    while not (entrada.isdigit() and 0 <= int(entrada) < len(ventas[0])):
        print("Ingrese un número de semana válido (0 a 3).")
        entrada = input("Ingrese el número de semana (0 a 3): ")
    return (entrada)

pedir_semana(ventas_semanales)
```
