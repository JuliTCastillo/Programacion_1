<p><b>a) ¿Qué significa que un objeto sea mutable o inmutable?</b></p>
<p>Que un objeto sea mutable quiere decir que su valor puede varia según el flujo del programa y un objeto inmutable guarda un valor fijo que no se puede modificar.</p>

<h4>Actividad 2</h4>
<p><b>a) ¿En qué se parecen una lista y una tupla?</p></b>
<p>Ambos pueden guardar una colección de datos ordenados y permiten acceder a ellos mediante indices (empenzando desde 0). Son secuencienciales, guarda datos de diferente tipo y son iterables.</p>
<p><b>b) ¿Cuál es la diferencia fundamental entre ambas?</p></b>
<p>Las <b>listas son mutables</b>, sus elementos pueden cambiar, modificarse y eliminarse. Por otro lado, las <b>tuplas</b> una vez que son creadas no se pueden modificar.</p>
<p><b>c) ¿Qué excepción produce intentar asignar un valor a una posición de una tupla?</p></b>
<p>Intentar modificar una tupla produce:
<code>TypeError: 'tuple' object does not support item assignment'</code></p>
<p><b>d) Mencionen un ejemplo de información que convendría representar mediante una tupla.</p></b>
<p>El uso de las tuplas debe usarse para guardar datos que no deben cambiar, por ejemplo: </br> Una fecha(día, mes, año) o meses("enero", "febrero",..., "diciembre")</p>

<h4>Actividad 3</h4>

```python
dias_habiles = ("Lunes", "Martes", "Miércoles", "Jueves", "Viernes")
tupla_vacia = ()
un_elemento = ("Python",)
registro = (1052, "Ana Pérez", 8.5)

print(type(dias_habiles))
print(type(tupla_vacia))
print(type(un_elemento))
print(type(registro))
print(type(("Python")))
print(type(("Python",)))
```

<p>En python, el comando <code>type(("python"))</code> lo toma como un <code>class 'str'</code> porque el que determina si es una tupla o no, es la coma que se ponen al final, por eso el comando <code>type(("Python",))</code> lo reconoce como <code>class 'tupla'</code></p>
