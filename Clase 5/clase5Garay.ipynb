<h4>1. Finalización anticipada de un ciclo: break</h4>
<p>a) Prueben con -3, 0, 1, 2, 9, 17 y 25.</p>

```python
def es_primo(numero):
    if numero < 2:
        return False

    esPrimo = True

    for i in range(2, numero):
        if (numero % i) == 0:
            esPrimo = False
            break
    return esPrimo
```

<p><b>b) Indiquen en qué caso se ejecuta break.</b></p>
<p>El break se ejecuta cuando se encuentra un divisor del número.</p>
<p><b>c) Expliquen por qué puede evitar iteraciones innecesarias.</b></p>
<p>Porque cuando encontramos un divisor del numero, ya sabemos que el número no es primo, por eso usamos el <code>break</code> que nos permite finalizar el ciclo antes de tiempo y evitar iteraciones innecesarias.</p>
<p><b>d) Comparen el resultado con una versión que recorra todos los posibles divisores.</b></p>

```python
def es_primo_sin_break(numero):
    if numero < 2:
        return False
    esPrimo = True
    for i in range(2, numero):
        if (numero % i) == 0:
            esPrimo = False
    return esPrimo
```

```python
def main():
    numerosPrueba = [ -3, 0, 1, 2, 9, 17, 25]
    for i in numerosPrueba:
        print(es_primo(i), end= " ")
    print("\n-->Usando la función sin break<--")
    for i in numerosPrueba:
        print(es_primo_sin_break(i), end= " ")
        

main()
```

<h4>2. Ciclos controlados con while True</h4>

```python
def carga_de_datos():
    listadoDeNumero = []
    while True:
        numero = int(input("Ingrese un número: "))
        if numero == -1: 
            break
        else:
            listadoDeNumero.append(numero)
    return listadoDeNumero

def main():
    listado = carga_de_datos()
    if len(listado) != 0:
        suma = sum(listado)
        cantidad = len(listado)
        print(f"--> Cantidad de datos: {cantidad}")
        print(f"--> Suma de datos: {suma}")
        print(f"--> Promedio de datos: {round(suma/cantidad, 2)}")
    else:
        print("No existen datos para calcular el promedio")

main()
```

<h4>3 - 4. Concepto de excepción y Captura con try-except</h4>

```python
try:
    numero = int(input("Ingrese un número: "))
    resultado = 100 / numero
except ZeroDivisionError:
    print("No se puede dividir por cero.")
except ValueError:
    print("Debe ingresar un valor numero entero.")
else:
    print(resultado)
```

<p>La diferencia de usar condicionales o try-except, es que los <b>condicionales</b> puede prevenir el error antes de hacer determinar instrucción pero el <b>try-except</b> lo que hace es capturar el error que se esta produciendo e informar que es lo que esta pasando, sin la necesidad que el programa explote.</p>

<h3>Actividad 5</h3>

```python
numeros = [10, 20, 30, 40]

try:
    posicion = int(input("Posición: "))
    if posicion < 0:
        raise NameError
    print(numeros[posicion])

except ValueError:
    print("No se ingresó un número invalido")

except NameError:
    print("La posición no puede ser negativa")

except IndexError:
    print(f"La posición no puede ser mayor a {len(numeros) - 1}")
```

<h4> Pruebas realizadas </h4>

| Entrada | Excepción capturada | Mensaje mostrado |
|---|---|---|
| "dos" | ValueError | "No se ingresó un número invalido" |
| 1 | (ninguna) | 20 |
| 8 | IndexError | "La posición no puede ser mayor a 3" |
| -1 | NameError | "La posición no puede ser negativa" |

<h5>¿Por qué conviene un mensaje distinto para cada problema?</h5>

Porque cada excepción indica una causa diferente del error, y un mensaje genérico 
("ocurrió un error") no le dice al usuario qué tiene que corregir. Si escribe letras, 
necesita saber que debe ingresar un número; si ingresa una posición fuera de rango, 
necesita saber cuál es el rango válido. Diferenciar los mensajes hace que el programa 
sea más fácil de usar y más fácil de depurar, porque cada `except` señala exactamente 
qué regla del problema no se cumplió.

<h3>Actividad 6</h3>

```python
try:
    a = int(input("Ingrese un numero: "))
    b = int(input("Ingrese un numero: "))
    division = a / b
except ValueError:
    print("Se ingreso un numero invalido")

except ZeroDivisionError:
    print("El segundo valor no puede ser 0")

else:
    print(f"El resultado de la division es {division}")

finally:
    print("Division finalizada")
```

<h3>Actividad 9</h3>

```python
def calcular_importe(cantidad, precio):
    if cantidad <= 0:
        raise ValueError("La cantidad debe ser mayor que cero.")
    if precio <= 0:
        raise ValueError("El precio debe ser mayor que cero.")

    importe = cantidad * precio

    # Comprobación interna del cálculo, no de la entrada
    assert importe == cantidad * precio, "Error interno en el cálculo del importe"

    return importe


def cargar_productos():
    productos = []

    while True:
        codigo = input("Código del producto (FIN para terminar): ")

        if codigo.upper() == "FIN":
            break

        descripcion = input("Descripción: ")

        try:
            cantidad = int(input("Cantidad: "))
            precio = float(input("Precio: "))

            importe = calcular_importe(cantidad, precio)
            producto = [codigo, descripcion, cantidad, precio]
            productos.append(producto)

            print(f"Producto cargado. Importe: ${importe:.2f}")

        except ValueError as error:
            print(f"Datos inválidos: {error}")
            print("El producto no fue incorporado.")

    return productos


def mostrar_resumen(productos):
    if not productos:
        print("No se cargaron productos.")
        return

    importe_total = sum(
        calcular_importe(producto[2], producto[3])
        for producto in productos
    )
    precio_promedio = sum(producto[3] for producto in productos) / len(productos)

    print("\n--- Resumen ---")
    print(f"Cantidad de productos: {len(productos)}")
    print(f"Importe total: ${importe_total:.2f}")
    print(f"Precio promedio: ${precio_promedio:.2f}")


def main():
    productos = cargar_productos()
    mostrar_resumen(productos)


main()
```
