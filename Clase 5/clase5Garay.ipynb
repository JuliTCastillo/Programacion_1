<h4>1. Finalización anticipada de un ciclo: break</h4>
<p>a) Prueben con -3, 0, 1, 2, 9, 17 y 25.</p>

```python
def es_primo(numero):
    if numero < 2:
        return False

    primo = True

    for i in range(2, numero):
        if numero % i == 0:
            primo = False
            break
    return primo
```

<p><b>b) Indiquen en qué caso se ejecuta break.</b></p>
<p>El break se ejecuta únicamente cuando el número evaluado es mayor o igual a 2 y se encuentra el primer divisor exacto en la condición <code>numero % i == 0</code></p>
<p><b>c) Expliquen por qué puede evitar iteraciones innecesarias.</b></p>
<p>Porque al detectar el primer divisor queda demostrado que el número no es primo [2]. La instrucción <code>break</code> interrumpe el ciclo de inmediato, evitando iteraciones restantes que no cambiarían el resultado final [1, 2]</p>
<p><b>d) Comparen el resultado con una versión que recorra todos los posibles divisores.</b></p>

```python
def es_primo_con_break(numero):
    if numero < 2:
        return False
    primo = True
    for i in range(2, numero):
        if numero % i == 0:
            primo = False
            break
    return primo
```

```python
def es_primo_sin_break(numero):
    if numero < 2:
        return False
    primo = True
    for i in range(2, numero):
        if numero % i == 0:
            primo = False
    return primo
```

<h4>2. Ciclos controlados con while True</h4>

```python
def solicitar_numeros():
    numeros = []
    while True:
        numero = int(input("Ingrese un número entero (-1 para terminar): "))
        if numero == -1:
            break
        numeros.append(numero)
    return numeros

def calcular_y_mostrar_resumen(numeros):
    if not numeros:
        print("No existen datos para calcular el promedio.")
        return

    cantidad = len(numeros)
    suma = sum(numeros)
    promedio = suma / cantidad

    print("\n--- Resumen de Datos ---")
    print(f"Cantidad de datos ingresados: {cantidad}")
    print(f"Suma de datos: {suma}")
    print(f"Promedio de datos: {promedio:.2f}")
    
    def main():
        lista_numeros = solicitar_numeros()
        calcular_y_mostrar_resumen(lista_numeros)

    main()
```

<h4>3 - 4. Concepto de excepción y Captura con try-except</h4>

```python
try:
    numero = int(input("Ingrese un número entero distinto de cero: "))
    resultado = 100 / numero
    print(f"El resultado de 100 / {numero} es: {resultado:.2f}")
    break  # Finaliza el ciclo solo si la conversión y división fueron exitosas
except ValueError:
    print("Error: El dato ingresado no es un número entero válido. Intente nuevamente.")
except ZeroDivisionError:
    print("Error: No se puede dividir por cero. Ingrese un valor diferente de 0.")
```

<p>La diferencia de usar condicionales o try-except, es que los <b>condicionales</b> puede prevenir el error antes de hacer determinar instrucción pero el <b>try-except</b> lo que hace es capturar el error que se esta produciendo e informar que es lo que esta pasando, sin la necesidad que el programa explote.</p>

<h3>Actividad 5</h3>

```python
numeros = [10, 20, 30, 40]

try:
    posicion = int(input("Posición: "))
    if posicion < 0:
        raise NameError
    print(numeros[posicion])

except ValueError:
    print("No se ingresó un número invalido")

except NameError:
    print("La posición no puede ser negativa")

except IndexError:
    print(f"La posición no puede ser mayor a {len(numeros) - 1}")
```

<h4>Resultados de las Pruebas (1, 8 y -1)</h4>
<ul>
    <li><b>Caso <code>"dos"</code>:</b> Lanza <code>ValueError</code> en la función <code>int()</code> al no ser una cadena numérica [1, 6].</li>
    <li><b>Caso <code>1</code>:</b> La conversión y el acceso son válidos, devolviendo el elemento <code>20</code> [1, 6].</li>
    <li><b>Caso <code>8</code>:</b> Lanza <code>IndexError</code> al no existir la posición 8 en la lista de 4 elementos [1, 6].</li>
    <li><b>Caso <code>-1</code>:</b> Aunque los índices negativos son sintácticamente válidos en Python, la validación explícita detecta que es menor a 0 y provoca un <code>ValueError</code> [1, 6].</li>
</ul>

<h5>¿Por qué conviene un mensaje distinto para cada problema?</h5>

<p>Porque cada excepción señala una causa diferente del error. Un mensaje genérico no le indica al usuario qué tiene que corregir: si ingresa texto, necesita saber que debe ingresar un número entero; si ingresa una posición inexistente o un índice negativo, debe comprender qué regla de la lista o del problema no se cumplió. Diferenciar los mensajes mediante bloques <code>except</code> específicos facilita la experiencia de uso del programa y permite al programador ubicar y depurar fallas rápidamente.</p>

<h3>Actividad 6</h3>

```python
try:
    num1 = float(input("Ingrese el dividendo (primer número): "))
    num2 = float(input("Ingrese el divisor (segundo número): "))
    resultado = num1 / num2
except ValueError:
    print("Error: Debe ingresar valores numéricos válidos.")
except ZeroDivisionError:
    print("Error: No se puede dividir por cero.")
else:
    print(f"El resultado de la división es: {resultado:.2f}")
finally:
    print("Intento de operación finalizado.")
```

<h3>Actividad 9</h3>

```python
def calcular_importe(cantidad, precio):
    if cantidad <= 0:
        raise ValueError("La cantidad debe ser mayor que cero.")
    if precio <= 0:
        raise ValueError("El precio debe ser mayor que cero.")

    importe = cantidad * precio

    # Comprobación interna del cálculo, no de la entrada
    assert importe == cantidad * precio, "Error interno en el cálculo del importe"

    return importe


def cargar_productos():
    productos = []

    while True:
        codigo = input("Código del producto (FIN para terminar): ")

        if codigo.upper() == "FIN":
            break

        descripcion = input("Descripción: ")

        try:
            cantidad = int(input("Cantidad: "))
            precio = float(input("Precio: "))

            importe = calcular_importe(cantidad, precio)
            producto = [codigo, descripcion, cantidad, precio]
            productos.append(producto)

            print(f"Producto cargado. Importe: ${importe:.2f}")

        except ValueError as error:
            print(f"Datos inválidos: {error}")
            print("El producto no fue incorporado.")

    return productos


def mostrar_resumen(productos):
    if not productos:
        print("No se cargaron productos.")
        return

    importe_total = sum(
        calcular_importe(producto[2], producto[3])
        for producto in productos
    )
    precio_promedio = sum(producto[3] for producto in productos) / len(productos)

    print("\n--- Resumen ---")
    print(f"Cantidad de productos: {len(productos)}")
    print(f"Importe total: ${importe_total:.2f}")
    print(f"Precio promedio: ${precio_promedio:.2f}")


def main():
    productos = cargar_productos()
    mostrar_resumen(productos)


main()
```
