## 1. Finalización anticipada de un ciclo: break

La instrucción `break` finaliza inmediatamente el ciclo `for` o `while` más cercano. La ejecución continúa en la primera instrucción ubicada después de ese ciclo. No finaliza por sí sola la función ni el programa completo.

Implementen `es_primo(numero)` utilizando una variable lógica y `break`. La función deberá retornar `False` para todo número menor que 2 y detener la búsqueda apenas encuentre un divisor.

**a) Prueben con -3, 0, 1, 2, 9, 17 y 25.**

```python
def es_primo(numero):
    if numero < 2:
        return False

    esPrimo = True

    for i in range(2, numero):
        if (numero % i) == 0:
            esPrimo = False
            break

    return esPrimo


casos = [-3, 0, 1, 2, 9, 17, 25]
for numero in casos:
    print(numero, "->", es_primo(numero))
```

**b) Indiquen en qué caso se ejecuta break.**

El `break` se ejecuta únicamente cuando el número es mayor o igual a 2 y NO es primo (en nuestros casos, 9 y 25), porque ahí es donde el ciclo encuentra un divisor exacto y corta la búsqueda. Con -3, 0 y 1 la función devuelve `False` antes de entrar al `for` (por el `if` inicial), así que el `break` nunca llega a ejecutarse. Con 2 y 17 el `for` recorre todos los divisores sin encontrar ninguno, así que el ciclo termina solo, sin `break`.

**c) Expliquen por qué puede evitar iteraciones innecesarias.**

Porque apenas se confirma que el número no es primo (se encontró un divisor), ya no hace falta seguir probando el resto de los divisores: el resultado no va a cambiar. Sin el `break`, el ciclo seguiría iterando hasta `numero-1` aunque la respuesta ya esté decidida, gastando iteraciones de más.

**d) Comparen el resultado con una versión que recorra todos los posibles divisores.**

El resultado final (`True`/`False`) es exactamente el mismo con o sin `break`. La diferencia está en la cantidad de iteraciones: para 25, con `break` se corta en el divisor 5 (iteraciones: 2, 3, 4, 5), mientras que sin `break` se sigue probando 6, 7, ..., 24 igual, aunque ya se sabía que no era primo.

```python
def es_primo_sin_break(numero):
    if numero < 2:
        return False

    esPrimo = True

    for i in range(2, numero):
        if (numero % i) == 0:
            esPrimo = False
            # sin break: sigue recorriendo aunque ya sepa que no es primo

    return esPrimo


for numero in [9, 25]:
    print(numero, "->", es_primo_sin_break(numero))
```

## 2. Ciclos controlados con while True

`while True` crea un ciclo cuya condición siempre es verdadera. Debe utilizarse solamente cuando la condición de finalización se comprueba dentro del bloque y conduce de manera clara a un `break`.

Desarrollen un programa que solicite números enteros y finalice cuando se ingrese -1. El valor -1 es la condición a controlar: controla la finalización y no debe incorporarse a los cálculos.

Al finalizar, informar cantidad, suma y promedio. Si el primer valor ingresado es -1, mostrar que no existen datos para calcular el promedio.

```python
cantidad = 0
suma = 0

while True:
    valor = int(input("Ingrese un número entero (-1 para terminar): "))

    if valor == -1:
        break

    cantidad += 1
    suma += valor

print("\n--- Resultado ---")
print("Cantidad de valores cargados:", cantidad)
print("Suma:", suma)

if cantidad == 0:
    print("No existen datos para calcular el promedio.")
else:
    promedio = suma / cantidad
    print("Promedio:", round(promedio, 2))
```

No se usó `while True` para reemplazar un `for` porque acá la cantidad de repeticiones no se conoce de antemano (depende de cuándo el usuario decida ingresar -1).

## 3. Concepto de excepción

Una excepción es una situación anormal detectada durante la ejecución del programa. Si no se captura, interrumpe el flujo normal y Python muestra un traceback con el tipo de excepción, el mensaje y la línea en la que se produjo.

Ejecuten el programa con 10, 0 y "hola". Identifiquen qué casos generan `ZeroDivisionError` y `ValueError`.

```python
numero = int(input("Ingrese un número: "))
resultado = 100 / numero
print(resultado)
```

**Identificación de los casos:**
- `numero = 10` → funciona bien, resultado = 10.0
- `numero = 0` → `ZeroDivisionError` (no se puede dividir por cero)
- `numero = "hola"` → `ValueError` (`int()` no puede convertir "hola"; este error ocurre antes de llegar a la división)

**Validación preventiva (con `if`) vs. captura correctiva (con `try-except`):** la preventiva anticipa el problema antes de que ocurra (por ejemplo, chequear `if numero != 0` antes de dividir), pero no sirve para el `ValueError` de `int()`, porque no se puede saber si un dato es convertible sin intentar convertirlo. La correctiva deja que el error ocurra y lo atrapa con `try-except`, permitiendo dar un mensaje claro y seguir ejecutando el programa en vez de cortarse con el traceback.

## 4. Captura con try-except

El bloque `try` contiene únicamente las instrucciones que pueden generar el error esperado. Cada `except` debe capturar una excepción específica.

Modifiquen el ejercicio del punto 3 para capturar `ValueError` cuando el usuario ingrese un dato no entero. El programa deberá informar el error y continuar solicitando valores sin perder los datos válidos ya cargados.

```python
valores_validos = []

while True:
    entrada = input("Ingrese un número entero (o 'fin' para terminar): ")

    if entrada.lower() == "fin":
        break

    try:
        numero = int(entrada)
    except ValueError:
        print("Error:", entrada, "no es un número entero válido. Los datos ya cargados no se pierden.")
        continue

    valores_validos.append(numero)
    print("Cargado correctamente. Total cargados:", len(valores_validos))

print("\n--- Valores válidos cargados ---")
print(valores_validos)
```

## 5. Múltiples excepciones y reglas del problema

Capturen `ValueError` cuando el dato no pueda convertirse a entero e `IndexError` cuando la posición quede fuera del rango de la lista. Recuerden que los índices negativos son válidos en Python; si la regla del problema solo permite posiciones entre 0 y `len(numeros)-1`, deberán validarlo expresamente y provocar un `ValueError`.

Prueben con "dos", 1, 8 y -1.

```python
numeros = [10, 20, 30, 40]

def obtener_elemento(lista, posicion_texto):
    try:
        posicion = int(posicion_texto)

        if posicion < 0:
            raise ValueError("La posición debe estar entre 0 y " + str(len(lista) - 1) + " (no se permiten índices negativos).")

        return lista[posicion]

    except ValueError as error:
        print("ValueError:", error)
    except IndexError:
        print("IndexError: la posición", posicion_texto, "está fuera del rango de la lista.")


for caso in ["dos", "1", "8", "-1"]:
    print("Posición ingresada:", caso)
    resultado = obtener_elemento(numeros, caso)
    if resultado is not None:
        print("Resultado:", resultado)
    print()
```

Conviene un mensaje diferente para cada problema porque cada excepción indica una causa distinta: "dos" no es un número (hay que ingresar dígitos), "-1" es un número válido para Python pero no cumple la regla del problema (hay que ingresar un valor dentro del rango permitido), y "8" es un número válido pero esa posición no existe en la lista (hay que ingresar una posición menor). Un mensaje genérico no permitiría entender ni corregir el problema real.

## 6. Cláusulas else y finally

Desarrollen un programa que solicite dos números y calcule su división. Utilicen `except ValueError` para datos no numéricos, `except ZeroDivisionError` para un divisor igual a cero, `else` para mostrar el resultado solamente si el `try` finalizó sin excepciones y `finally` para mostrar un mensaje de cierre que se ejecute exista o no un error.

```python
try:
    numerador = float(input("Ingrese el numerador: "))
    denominador = float(input("Ingrese el denominador: "))
    resultado = numerador / denominador

except ValueError:
    print("Error: ambos valores deben ser numéricos.")

except ZeroDivisionError:
    print("Error: el divisor no puede ser cero.")

else:
    print("Resultado:", resultado)

finally:
    print("Intento de división finalizado.")
```

El bloque `else` no reemplaza al `try`: solo separa el código que depende de que todo haya salido bien (mostrar el resultado), para que no quede mezclado dentro del `try`. `finally` no indica que la operación fue exitosa: se ejecuta siempre, haya o no error, y solo garantiza la ejecución de tareas finales, como cerrar un recurso o informar que el intento terminó.

## 7. Provocar una excepción con raise

`raise` permite señalar que un dato no cumple las reglas del problema, aun cuando su tipo sea correcto. Implementen `calcular_importe(cantidad, precio)` de modo que provoque `ValueError` si `cantidad` o `precio` son menores o iguales a cero. La función deberá retornar `cantidad * precio` cuando los datos sean válidos.

Desde el programa principal, capturen la excepción y muestren su mensaje utilizando `except ValueError as error`. Prueben con datos válidos, cantidad cero, precio negativo y entradas no numéricas.

```python
def calcular_importe(cantidad, precio):
    if cantidad <= 0:
        raise ValueError("La cantidad debe ser mayor que cero.")
    if precio <= 0:
        raise ValueError("El precio debe ser mayor que cero.")

    return cantidad * precio


def probar_caso(cantidad_texto, precio_texto):
    print("cantidad =", cantidad_texto, "| precio =", precio_texto)
    try:
        cantidad = int(cantidad_texto)
        precio = float(precio_texto)
        importe = calcular_importe(cantidad, precio)
    except ValueError as error:
        print("ValueError:", error)
    else:
        print("Importe:", importe)
    print()


probar_caso("3", "1200")
probar_caso("0", "1200")
probar_caso("3", "-1200")
probar_caso("tres", "1200")
```

## 8. Comprobaciones internas con assert

`assert` comprueba una condición que el programador espera que siempre sea verdadera. Si resulta falsa, genera `AssertionError`. Es una ayuda para detectar errores de lógica o verificar invariantes internas; no debe utilizarse como mecanismo principal para validar datos ingresados por el usuario, ya que las aserciones pueden desactivarse.

```python
def calcular_importe(cantidad, precio):
    if cantidad <= 0:
        raise ValueError("La cantidad debe ser mayor que cero.")
    if precio <= 0:
        raise ValueError("El precio debe ser mayor que cero.")
    return cantidad * precio


importe = calcular_importe(3, 1200)
assert importe == 3600, "El importe calculado no es el esperado"
print("OK:", importe)
```

Escriban otras dos aserciones: una que verifique un caso correcto y otra que falle intencionalmente.

```python
# Aserción que verifica un caso correcto
importe_2 = calcular_importe(5, 200)
assert importe_2 == 1000, "El importe calculado no es el esperado"
print("OK:", importe_2)

# Aserción que falla intencionalmente (2 * 100 = 200, no 999)
importe_3 = calcular_importe(2, 100)
assert importe_3 == 999, "El importe calculado no es el esperado"
print("OK:", importe_3)  # esta línea nunca se imprime
```

**Interpretación del AssertionError obtenido:** las dos primeras aserciones pasan sin problema. La tercera falla porque `calcular_importe(2, 100)` da 200, no 999: Python corta la ejecución ahí mismo y muestra `AssertionError: El importe calculado no es el esperado`. Esto confirma que `assert` no informa amablemente el problema como haría un `except`: interrumpe el programa directamente, porque asume que si la condición es falsa hay un error de lógica en el código, no un dato incorrecto del usuario.

## 9. Desafío integrador: carga segura de productos

Desarrollen un sistema modular que registre productos en una lista. Cada producto podrá almacenarse como una lista con código, descripción, cantidad y precio. La carga finaliza cuando el código ingresado sea "FIN". El programa deberá utilizar `while True` y `break` para finalizar la carga, capturar `ValueError` al convertir cantidad y precio, usar `raise ValueError` para rechazar cantidades o precios menores o iguales a cero, no incorporar un producto cuando sus datos sean inválidos, calcular el importe mediante una función, utilizar `assert` para comprobar una condición interna del cálculo, e informar cantidad de productos, importe total y precio promedio evitando la división por cero.

```python
def calcular_importe(cantidad, precio):
    if cantidad <= 0:
        raise ValueError("La cantidad debe ser mayor que cero.")
    if precio <= 0:
        raise ValueError("El precio debe ser mayor que cero.")

    importe = cantidad * precio
    assert importe == cantidad * precio, "Error interno en el cálculo del importe"
    return importe


def solicitar_producto():
    codigo = input("Código del producto ('FIN' para terminar): ")

    if codigo.upper() == "FIN":
        return "FIN"

    descripcion = input("Descripción: ")

    try:
        cantidad = int(input("Cantidad: "))
        precio = float(input("Precio: "))
        importe = calcular_importe(cantidad, precio)
    except ValueError as error:
        print("Producto rechazado:", error, "\n")
        return None

    print("Producto cargado. Importe:", importe, "\n")
    return [codigo, descripcion, cantidad, precio, importe]


productos = []

while True:
    producto = solicitar_producto()

    if producto == "FIN":
        break

    if producto is None:
        continue

    productos.append(producto)

cantidad_productos = len(productos)

print("\n--- Resumen de la carga ---")
print("Cantidad de productos cargados:", cantidad_productos)

if cantidad_productos == 0:
    print("No se cargaron productos: no hay importe total ni precio promedio.")
else:
    importe_total = sum(p[4] for p in productos)
    precio_promedio = sum(p[3] for p in productos) / cantidad_productos
    print("Importe total:", importe_total)
    print("Precio promedio:", round(precio_promedio, 2))
```
