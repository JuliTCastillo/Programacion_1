<h4>2. Metacaracteres, clases y cuantificadores</h4>

<p>a) <code>^ y $</code>: <code>^</code> indica como debe iniciar el texto y <code>$</code> como debe terminar el texto. <br><br>Ej: "El perro corre" <code>^El</code> se encuentra coincidencia en el texto y para el final <code>$rre</code> hace referencia al final del texto.</p>
<p>b) <code>.</code>: El punto representa cualquier carectar de manera individual (letra, número, espacio, símbolo), excepto una nueva línea.
<br><br>
Ej: c.sa coincide con "casa", "cesa" o "cosa", pero no con "csa"</p>
<p>c) <code>[]</code>: Los corchete nos sirve para indicar un determinado rango como con numeros [0-9] o letras [a-z]
<br><br>
Ej: [0-9] coincide con cualquier dígito individual, como el "5" en "Tengo 5 gatos". [A-Z] coincide con cualquier letra mayúscula, como la "M" en "Mi casa".
</p>
<p>d) <code>[^]</code>: Cuando tenemos <code>^</code> se representa como negación, se coloca al inicio. A diferencia de los anteriores que hacen referencia a "coincide con estos caracteres", en este caso es "<b>coincide con cualquier carácter que NO esté en esta lista</b>".

Ej: [^0-9] coincide con cualquier carácter que no sea un dígito, como la "a" en "a1b2".</p>
<p>e) <code>|</code>: Se usa para aclarar alternativas, funcionando como un "o lógico": permite que el patrón coincida con cualquiera de las opciones separadas por la barra vertical.
<br><br>
Ej: gato|perro coincide tanto con "gato" como con "perro" dentro de un texto.</p>
<p>f) <code>()</code>: Se usa para agrupar patrones para tratarla como una unidad, lo que nos permite aplicarle cuantificadores o el operador | solo a ese subconjunto, en lugar de a toda la expresión.

Ej: (ab)+ coincide con "ab", "abab", "ababab", etc. (el grupo completo "ab" se repite), mientras que sin paréntesis ab+ solo repetiría la "b" ("a", "ab", "abb", "abbb"...).</p>
<p>g) <code>?, * y +</code>: nos ayuda con repeticiones del elemento anterior. Esto son cuantificadores que se aplican al carácter o grupo que tienen justo antes:
<br><br>
? → el elemento puede aparecer 0 o 1 vez (es opcional).<br><br>
* → el elemento puede aparecer 0 o más veces.<br><br>
+ → el elemento debe aparecer al menos 1 vez, pero puede repetirse más.<br><br>

Ej: colou?r coincide con "color" y "colour". ab*c coincide con "ac", "abc", "abbc". ab+c coincide con "abc", "abbc", pero no con "ac".</p>
<p>h) <code>{m}, {m,n} y {m,}</code>: Expresa la cantidad exacta o rango de repeticiones, las llaves permiten especificar con precisión cuántas veces debe repetirse el elemento anterior:
<br><br>
{m} → exactamente m veces.
{m,n} → entre m y n veces.
{m,} → al menos m veces, sin límite superior.
<br><br>
Ej: [0-9]{4} coincide con exactamente 4 dígitos, como "2026". [0-9]{2,4} coincide con secuencias de 2 a 4 dígitos. a{3,} coincide con "aaa", "aaaa", "aaaaa", etc.</p>
<p>i) <code>\</code>: Esto nos ayuda a separar los significado de los metacaracteres con los simbolos especiales.
<br><br>
Ej: si querés buscar literalmente un signo de pesos, usás \$, porque $ solo (sin escapar) significaría "final de la cadena". Así, \$100 coincide con el texto "$100" tal cual aparece.</p>

<h4>3. Análisis y construcción de patrones</h4>
<code>r"^PROG[0-9]{3}$"</code> : <p>El patron busca un "código" que empiece con <b>PROG</b> que tenga 3 número aleatorio que este de un rango de 0 a 9. Ej: <b>PROG321</b></p>
<code>r"^[A-Za-z]{3,10}$"</code> : <p>Se busca un palabra cualquiera que tenga entre 3 o 10 letras (mayúscula o mínuscula). Ej: Anátomia</p>
<code>r"^[0-9]{4}-[0-9]{4}$"</code> : <p>Busca un patron que tenga números de 4 digitos al inicio y al final. Ej: 1990-2010</p>
<code>r"^([A-Z]{3}[0-9]{3}|[A-Z]{2}[0-9]{3}[A-Z]{2})$</code> : <p>Tenemos dos alternativas de patrones, busca un patros de 2 o 3 letras mayusculas con 2 o 3 digitos. Ej: AU390 o DT43.</p>

<hr>
<p>a) Un código formado por dos letras mayúsculas y cuatro números.</p>
<code>r"^[A-Z]{2}[0-9]{4}"</code>
<p>b) Un legajo de exactamente cinco dígitos.</p>
<code>r"^[0-9]{5}"</code>
<p>c) Un importe formado por el símbolo $ y uno o más dígitos. Recuerden escapar el símbolo: \$.</p>
<code>r"^\$[0-9]+"</code>
<p>d) Una palabra que comience con mayúscula y continúe únicamente con letras.</p>
<code>[A-Z][a-zA-Z]*</code>
<p>e) Una fecha con formato DD/MM/AAAA, validando solamente la forma y no la existencia real de la fecha.</p>
<code>r"[0-9]{2}/[0-9]{2}/[0-9]{4}"</code>

<h4>4. match, search y fullmatch</h4>

````python
import re
patron = r"[A-Za-z][0-9]{3}"
codigo = "A123XYZ"
inicio = re.match(patron, codigo)
completo = re.fullmatch(patron, codigo)

print(inicio)
print(completo)
````

<p>a) El <code>re.match()</code> solo verifica que el patrón coincida <i>a partir del inicio</i> de la cadena, pero no exige que coincida con TODO EL TEXTO. En este caso, "A123" cumple con el patrón solicitado pero luego no, match se guarda la parte que coincide por lo que encuentra un resultado.</p>
<p>b) <code>re.fullmatch()</code> busca que <i>TODO EL TEXTO</i> cumpla con el patrón ingresado. Como el código completo no cumple con el patrón solo una parte, <code>re.fullmatch()</code> lo toma como que no encontro coinicidencia.</p>

<p>Dado un texto con datos heterogéneos, el módulo <code>re</code> permite buscar todas las apariciones de un patrón mediante <code>findall()</code> o <code>finditer()</code>.</p>

<pre><code>import re </code></pre>

texto = "Game Hub- Correo: gamehub@gmail.com - Tel: 1234-5678 - Código: GH2026. Lucas Garay - Correo: lgaray@uade.edu.ar - Tel: 8765-4321 - Código: LG6202."

<h5> a) re.findall() para obtener todas las secuencias numéricas</h5>
<pre><code>numeros = re.findall(r"+", texto)
print("a) Secuencias numéricas encontradas con findall():")

<h5> b) re.finditer() para recorrer coincidencias e informar contenido y posiciones </h5>
<pre><code>print("\nb) Coincidencias e índices detallados con finditer():")

<p><b><code>re.findall()</code>:</b> Devuelve directamente una lista de cadenas de texto con los valores encontrados, o una lista de tuplas si la expresión regular contiene grupos de captura.</p>
<p><b><code>re.finditer()</code>:</b> Devuelve un iterador que produce objetos <code>Match</code>. Cada objeto permite consultar información detallada sobre la coincidencia mediante métodos como <code>start()</code>, <code>end()</code> y <code>span()</code>.</p>

<h3>Punto 7: Funciones de validación</h3>
<p>Implementación de funciones de validación utilizando <code>re.fullmatch()</code> para verificar el formato completo de las cadenas.</p>

<pre><code>import re

def validar_legajo(legajo):
    return re.fullmatch(r"{5}", legajo) is not None

def validar_codigo(codigo):
    return re.fullmatch(r"[A-Z]{2}{4}", codigo) is not None

def validar_telefono(telefono):
    return re.fullmatch(r"{4}-{4}", telefono) is not None

def validar_correo(correo):
    return re.fullmatch(r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}", correo) is not None

def validar_importe(importe):
    return re.fullmatch(r"\$+", importe) is not None

def validar_patente(patente):
    return re.fullmatch(r"^([A-Z]{3}{3}|[A-Z]{2}{3}[A-Z]{2})$", patente) is not None
</code></pre>
