```python
grupo_a = {101, 102, 103, 104, 105}
grupo_b = {104, 105, 106, 107}
```

<p>Resolver e interpretar cada resultado:</p>
<p><b>a) Unión: grupo_a | grupo_b.</b></p>
<p>Combinas los valores de ambos conjuntos sin repeticiones.</p>

```python
grupo_c = grupo_a | grupo_b
print(grupo_c)
```

<p><b>b) Intersección: grupo_a & grupo_b.</b></p>
<p>Nos muestra los valores que se encuentran en ambos conjuntos.</p>

```python
grupo_c = grupo_a & grupo_b
print(grupo_c)
```

<p><b>c) Diferencia: grupo_a - grupo_b y grupo_b - grupo_a.</b></p>
<p>La resta simétrica, nos devuelve los valores que no se encuentran en el primer conjunto.</p>

```python
grupo_c = grupo_a - grupo_b
print(grupo_c)
grupo_d = grupo_b - grupo_a
print(grupo_d)
```

<p><b>d) Diferencia simétrica: grupo_a ^ grupo_b.</b></p>
<p>Nos devuelve un conjuntos excluyendo los datos que aparecen en ambos conjuntos</p>

```python
grupo_c = grupo_a ^ grupo_b
print(grupo_c)
```

<p><b>e) Pertenencia: 103 in grupo_a y 108 not in grupo_b</b></p>
<p>Estas dos expresiones nos sirve para saber si dicho valor se encuentra en el conjunto o si no se encuentra, estos devuelve un true o false.`</p>

```python
103 in grupo_a
108 not in grupo_b
```

<p><b>f) Explicar por qué la diferencia no es una operación conmutativa.</b></p>
<p>Esto es porque la expresión depende del primer conjunto, por ejemplo: </br><b>grupo_a - grupo_b</b> Se queda con los elementos del primer conjunto, <b>excluyendo</b> lo que que también están en el segungo.</p>

## Ej. 5

```python
alumnos = {
    1001: "Ana",
    1002: "Bruno",
    1003: "Carla"
}
diccionario_vacio = {}

print(alumnos)
print(type(diccionario_vacio))
print(type(set()))
```

**a) ¿Qué representa cada clave y cada valor?**
Cada clave es el legajo de un alumno (1001, 1002, 1003) y funciona como identificador. Cada valor es el nombre del alumno asociado a ese legajo.

**b) ¿Qué ocurriría si se volviera a asignar un valor a la clave 1002?**
Como las claves son únicas, se reemplazaría el valor anterior.

**c) ¿Por qué una lista no puede utilizarse como clave?**
Porque las claves deben ser inmutables. El contenido de una lista puede cambiar, y si cambiara la clave, el diccionario ya no podría ubicar el valor asociado.

**d) ¿Qué diferencia existe entre `{}` y `set()`?**
`{}` crea un diccionario vacío, mientras que `set()` crea un conjunto vacío. Ambos usan llaves cuando tienen contenido, pero el diccionario guarda pares `clave: valor` y el conjunto solo guarda elementos únicos. Por eso el conjunto vacío no puede escribirse como `{}`.

<h3>Ejercicio 7</h3>

```python
stock = {
    "teclado": 12,
    "mouse": 20,
    "monitor": 7,
    "webcam": 5
}

# a) Recorrer directamente el diccionario y mostrar sus claves
print("a) Claves (productos):")
for producto in stock:
    print(producto)

# b) Mostrar las cantidades mediante values()
print("\nb) Cantidades (valores):")
for cantidad in stock.values():
    print(cantidad)

# c) Mostrar cada producto y su stock mediante items()
print("\nc) Producto y stock:")
for producto, cantidad in stock.items():
    print(f"{producto}: {cantidad} unidades")

# d) Informar la cantidad de productos diferentes con len()
print(f"\nd) Productos diferentes: {len(stock)}")

# e) Calcular el total de unidades disponibles
total_unidades = sum(stock.values())
print(f"\ne) Total de unidades disponibles: {total_unidades}")

# f) Informar el producto con mayor stock mediante max(stock, key=stock.get)
producto_mas_stock = max(stock, key=stock.get)
print(f"\nf) Producto con mayor stock: {producto_mas_stock} ({stock[producto_mas_stock]} unidades)")
```

## Ejercicio 9

```python
# Actividad integradora: análisis de una frase
frase = input("Ingresá una frase: ")

# a) Normalizar a minúsculas
frase = frase.lower()

# b) Eliminar signos de puntuación (se reemplazan por espacio para no pegar palabras)
signos = ".,;:¿?¡!"
for signo in signos:
    frase = frase.replace(signo, " ")

# c) Separar en palabras (split() ignora espacios repetidos)
palabras = frase.split()

# d) Conjunto de palabras diferentes
palabras_unicas = set(palabras)

# e) Diccionario de frecuencias
frecuencias = {}
for palabra in palabras:
    frecuencias[palabra] = frecuencias.get(palabra, 0) + 1

# f) Palabras sin repetir en orden alfabético
print("\nf) Palabras sin repetir (orden alfabético):")
for palabra in sorted(palabras_unicas):
    print("-", palabra)

# g) Cada palabra con su frecuencia
print("\ng) Frecuencia de cada palabra:")
for palabra, cantidad in frecuencias.items():
    print(f"{palabra}: {cantidad}")

# h) Palabra más frecuente (solo si se ingresó al menos una)
if frecuencias:
    mas_frecuente = max(frecuencias, key=frecuencias.get)
    print(f"\nh) Palabra más frecuente: {mas_frecuente} ({frecuencias[mas_frecuente]} veces)")
else:
    print("\nh) No se ingresó ninguna palabra.")
```
