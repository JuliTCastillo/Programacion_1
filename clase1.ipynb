Algoritmo MultaBiblioteca

    Definir fechaLimite, fechaDevolucion Como Fecha
    Definir dias Como Entero
    Definir multa Como Real
    Definir reservado Como Cadena

    Escribir "Ingrese la fecha límite:"
    Leer fechaLimite

    Escribir "Ingrese la fecha de devolución:"
    Leer fechaDevolucion

    Escribir "¿El libro estaba reservado? (S/N):"
    Leer reservado

    dias <- DiferenciaEnDias(fechaDevolucion, fechaLimite)

    Si dias < 0 Entonces

        Escribir "Error: la cantidad de días no puede ser negativa."

    Sino

        multa <- dias * 800

        Si dias >= 1 Y reservado = "S" Entonces
            multa <- multa + 2500
        FinSi

        Si multa = 0 Entonces
            Escribir "No corresponde multa."
        Sino
            Escribir "La multa total es: $", multa
        FinSi

    FinSi

FinAlgoritmo

```python
dias = int(input("Ingrese los días de atraso: "))
reservado = input("¿El libro estaba reservado? (S/N): ").upper()

if dias < 0:
    print("Error: la cantidad de días no puede ser negativa.")
else:
    multa = dias * 800

    if dias >= 1 and reservado == "S":
        multa += 2500

    if multa == 0:
        print("No corresponde multa.")
    else:
        print(f"La multa es de ${multa}")
```
