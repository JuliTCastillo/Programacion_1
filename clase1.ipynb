Implementen y documenten las siguientes funciones:<p>

<hr>
<p>d) Las funciones que <b>retornar un valor</b> es para poder almacenar el valor y seguir usandolo en el programa. En cambio, cuando nosotros <b>mostramos un valor por pantalla</b> es más que nada para informar al usuario y el valor se pierde.</p>

```python
def calcular_multa(diaDeAtraso, reservado):
    if diaDeAtraso < 0:
            print("Error: la cantidad de días no puede ser negativa.")
    else:
        multa = diaDeAtraso * 800 #Agrega multa por no entregar a tiempo

        #Verificamos si el libro estaba reservado
        if diaDeAtraso >= 1 and reservado == "S":
            multa += 2500 #Agregamos la recarga
    
        if multa == 0:
            return "No corresponde multa."
        else:
            return f"La multa es de ${multa}"

def mostrar_resultado(nombreUsuario, mensaje):
    print(nombreUsuario, mensaje)

def main():
    #Estructura de la lista [dia, mes, año]
    fechaLimite = [8, 8, 27]
    fechaRegreso = [10, 8, 27]
    dias = fechaRegreso[0] - fechaLimite[0]
    nombre = input("Ingrese su nombre").title()
    reservado = input("¿El libro estaba reservado? (S/N): ").upper()

    if reservado == "S": 
        reservado = True
    else:
        reservado = False

    respuesta = calcular_multa(dias, reservado)
    mostrar_resultado(nombre, respuesta)

main()
```

<p>Registren en una lista las multas correspondientes a 10 usuarios. Para cada usuario, soliciten los datos

Luego implementen funciones que permitan:</p>

<p>Verifiquen que las funciones reciban la lista como parámetro y que cada una resuelva una única 
responsabilidad.</p>

```python
def cargaDatos(listaMulta, listaNombre):
    for i in range(10):
        nombre = input("Ingrese su nombre: ")
        listaNombre.append(nombre)
        multa = int(input("Ingrese el monto de la multa: "))
        listaMulta.append(multa)

def mayorMulta(listaMulta):
    max = -9999
    for i in range(len(listaMulta)):
        if listaMulta[i] > max:
            max = listaMulta[i]
    return max

def calcularMulta(listaMulta):
    contador = 0
    for i in range(len(listaMulta)):
        contador += listaMulta[i]

    return contador / len(listaMulta)

def superarLimite(listaMulta, limite):
    contador = 0
    for i in range(len(listaMulta)):
        if listaMulta[i] > limite:
            contador += 1
    print(contador, "multas superan el limite.")

def sinMulta(listaMulta):
    contador = 0
    for i in range(len(listaMulta)):
        if listaMulta[i] == 0:
            contador += 1
    print(contador, "usuarios no tiene multas!")
 
def main():
    listaMulta = []
    listaNombre = []
    cargaDatos(listaMulta, listaNombre)
    print("La multa mayor fue de ", mayorMulta(listaMulta))
    print("El promedio de la lista de la multa es:", calcularMulta(listaMulta))
    superarLimite(listaMulta, 1000)
    sinMulta(listaMulta)

main()
```

información mediante una matriz de 4 filas y 7 columnas: cada fila corresponde a una sucursal y cada columna a un día.<br>
Creen la matriz con filas independientes. Eviten replicar una lista completa, porque todas las filas podrían referenciar el mismo objeto.</p>

```python
def generarMatriz(matriz):
    for i in range(4):
        fila = []
        for j in range(4):
            dato = int(input("Ingrese valores mayores a 0:"))
            while dato < 0:
                dato = int(input("Error. Ingrese valores mayores a 0:"))
            fila.append(dato)
        
        matriz.append(fila)

def recaudacionPorSucursal(matriz):
    contador = 0
    for i in range(len(matriz)):
        for j in range(len(matriz)):
            contador += matriz[i][j]
        print("La surcursal",i+1,"recaudo esta semana: $", contador)
        contador = 0

def recaudacionPorDia(matriz):
    dias = ["lunes", "martes", "miercoles", "jueves", "viernes", "sabado","domingo"]
    for i in range(len(matriz)):
        print("\nLa surcursal", i+1)
        for j in range(len(matriz)):
            print(dias[j].title(), ":", matriz[i][j]," ", end="")

def recaudacionGeneral(matriz):
    contador = 0
    for i in range(len(matriz)):
        for j in range(len(matriz)):
            contador += matriz[i][j]
    print("\n\nEn total se recaudo: $", contador, "\n")

def mayorRecaudacion(matriz):
    max = -9999
    numDia = 0
    dias = ["lunes", "martes", "miercoles", "jueves", "viernes", "sabado","domingo"]
    for i in range(len(matriz)):
        for j in range(len(matriz)):
            if max < matriz[i][j]:
                max = matriz[i][j]
                numDia = i
    print("La mayor recaudacion fue de", max, "el día", dias[numDia])

def main():
    matriz = []
    generarMatriz(matriz)
    recaudacionPorSucursal(matriz)
    recaudacionPorDia(matriz)
    recaudacionGeneral(matriz)
    mayorRecaudacion(matriz)
main()
```
